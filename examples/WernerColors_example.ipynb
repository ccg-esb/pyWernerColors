# WernerColors for Python

A simple example of how to use the Python implementation of **WernerColors**.

The package provides Werner color palettes, Matplotlib colormaps, custom palettes built from Werner color names, and nearest-color matching.

## Import

```python
import numpy as np
import matplotlib.pyplot as plt
import wernercolors as wc
```

## Available palettes

List the palettes included in the package.

```python
wc.available_palettes()
```

## Get colors from a palette

`palette()` returns colors as hexadecimal values.

```python
wc.palette("Firebirds")
```

You can request a specific number of colors.

```python
wc.palette("Firebirds", n=5)
```

Use `direction=-1` to reverse a palette.

```python
wc.palette("Firebirds", n=5, direction=-1)
```

## Visualize a palette

`plot_werner_strip()` provides a quick preview of a Werner palette.

```python
wc.plot_werner_strip("Rocks")
```

```python
wc.plot_werner_strip(
    "Firebirds",
    n=12,
    title="Werner: Firebirds"
)
```

## Continuous colormaps

Werner palettes can also be used as continuous Matplotlib colormaps.

```python
data = np.random.default_rng(1).random((20, 20))

plt.imshow(data, cmap=wc.cmap("Bugs"))
plt.colorbar(label="Value")
plt.show()
```

## Use a Werner palette for plot colors

`set_color_cycle()` sets the Matplotlib color cycle for an axis.

```python
x = np.linspace(0, 10, 200)

fig, ax = plt.subplots()

wc.set_color_cycle(ax, "Firebirds", n=4)

for i in range(4):
    ax.plot(x, np.sin(x + i), label=f"Series {i + 1}")

ax.legend()
plt.show()
```

## Build a palette from Werner color names

Individual Werner colors can be selected by name and combined into a custom palette.

```python
labels = [
    "Campanula Purple",
    "Vermilion Red",
    "Siskin Green",
    "Leek Green",
]

colors = wc.palette_from_names(labels)

colors
```

```python
wc.plot_palette_strip(
    colors,
    labels=labels,
    title="Custom Werner palette",
    auto_contrast=True,
)
```

## Find the closest Werner color

`closest_color()` finds the Werner color closest to an arbitrary hexadecimal color.

```python
wc.closest_color("#53658f")
```

You can also return several closest matches.

```python
wc.closest_color("#53658f", top_k=3)
```

## R-style interface

For users familiar with the original R package, `werner_brewer()` provides a similar interface.

```python
wc.werner_brewer(
    "Firebirds",
    n=7,
    type="discrete",
    direction=1,
    return_hex=True,
)
```

## Summary

The main Python functions are:

- `wc.palette()` for discrete palettes
- `wc.cmap()` for continuous Matplotlib colormaps
- `wc.plot_werner_strip()` for palette previews
- `wc.palette_from_names()` for custom palettes
- `wc.closest_color()` for Werner color matching
